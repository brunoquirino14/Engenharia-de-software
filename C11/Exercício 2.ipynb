**Questão 1**

```python
nome_completo = input("Digite seu nome completo: ").strip()

print(f"Maiúsculas: {nome_completo.upper()}")

print(f"Minúsculas: {nome_completo.lower()}")

total_letras = len(nome_completo.replace(" ", ""))
print(f"Total de letras do nome: {total_letras}")

primeiro_nome = nome_completo.split()[0]
nome_alterado = f"{primeiro_nome} do Inatel"

print(f"Nome alterado: {nome_alterado}")
```

**Questão 2**

```python
numero = int(input("Deseja ver a tabuada de qual número? "))

inicio = int(input("A tabuada deve começar em qual número? "))
fim = int(input("A tabuada deve terminar em qual número? "))

print(f"\n--- Tabuada do {numero} (de {inicio} até {fim}) ---")

for i in range(inicio, fim + 1):
    resultado = numero * i
    print(f"{numero} x {i} = {resultado}")
```

**Questão 3**

```python
sexo = input("Digite o sexo (M ou F): ").upper()

while sexo != 'M' and sexo != 'F':
    sexo = input("Valor inválido! Por favor, digite M ou F: ").upper()

if sexo == 'M':
    print("Homem")
else:
    print("Mulher")
```

**Questão 4**

```python
distancia = float(input("Qual a distância da viagem em Km? "))

if distancia <= 200:
    preco = distancia * 0.50
else:
    preco = distancia * 0.45

print(f"O preço da passagem será de R$ {preco:.2f}")
```

**Questão 5**

```python
while True:
    numero_str = input("Digite um número entre 1000 e 9999: ").strip()

    if numero_str.isdigit() and 1000 <= int(numero_str) <= 9999:
        break
    else:
        print("Valor inválido! O número deve estar entre 1000 e 9999.")

print(f"Unidade: {numero_str[3]}")
print(f"Dezena: {numero_str[2]}")
print(f"Centena: {numero_str[1]}")
print(f"Milhar: {numero_str[0]}")
```

**Questão 6**

```python
import math

numero = float(input("Digite um número decimal: "))

print(f"Raiz quadrada: {math.sqrt(numero)}")
print(f"Função teto: {math.ceil(numero)}")
print(f"Função chão: {math.floor(numero)}")
print(f"Parte inteira: {math.trunc(numero)}")
```

**Questão 8**

```python
n1 = float(input("Digite o primeiro número: "))
n2 = float(input("Digite o segundo número: "))

adicao = n1 + n2

print(f"Adição: {adicao}")
print(f"Subtração: {n1 - n2}")
print(f"Multiplicação: {n1 * n2}")
print(f"Divisão: {n1 / n2}")
print(f"Resto da divisão: {n1 % n2}")
print(f"Potência: {n1 ** n2}")

if adicao % 2 == 0:
    print("O resultado da adição é par.")
else:
    print("O resultado da adição é ímpar.")
```
