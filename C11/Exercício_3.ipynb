```python
times = ['Real Madrid', 'Girona', 'Barcelona', 'Atlético de Madrid', 'Athletic Bilbao']

print(f"Os 3 primeiros colocados: {times[:3]}")

print(f"Os 2 últimos colocados: {times[-2:]}")

print(f"Times em ordem alfabética: {sorted(times)}")

posicao_barcelona = times.index('Barcelona') + 1
print(f"O Barcelona se encontra na {posicao_barcelona}ª posição.")
```

```python
loja_A = {'iPhone 6', 'Galaxy S25', 'Moto G54', 'Poco X5'}
loja_B = {'iPhone 14', 'Galaxy S23', 'Moto G54', 'Redmi Note 12'}

total_modelos = loja_A | loja_B
print(f"Opções totais de modelos ao visitar as duas lojas:\n{total_modelos}\n")

modelos_em_comum = loja_A & loja_B
print(f"Modelos disponíveis em ambas as lojas:\n{modelos_em_comum}")
```

### Questão 3


```python
aluno = {}
aluno['nome'] = input("Digite o nome do aluno: ")
aluno['media'] = float(input("Digite a média do aluno: "))

if aluno['media'] >= 50:
    aluno['Situação:'] = 'AP'
else:
    aluno['Situação:'] = 'RP'

print("\nDados do Dicionário:")
print(aluno)

print("\nRelatório Final:")
for chave, valor in aluno.items():
    print(f"{chave.capitalize()}: {valor}")
```

### Questão 4

```python
pessoas = []

for i in range(3):
    nome = input(f"Digite o nome da {i+1}ª pessoa: ")
    peso = float(input(f"Digite o peso de {nome} (kg): "))
    pessoas.append({'nome': nome, 'peso': peso})

mais_pesada = pessoas[0]
mais_leve = pessoas[0]

for p in pessoas:
    if p['peso'] > mais_pesada['peso']:
        mais_pesada = p
    if p['peso'] < mais_leve['peso']:
        mais_leve = p

print(f"\nA pessoa mais pesada é {mais_pesada['nome']} com {mais_pesada['peso']}kg.")
print(f"A pessoa mais leve é {mais_leve['nome']} com {mais_leve['peso']}kg.")
```

### Questão 5

```python
n = int(input("Deseja cadastrar quantas pessoas? "))

soma_idade = 0
mulheres_menos_20 = 0

for i in range(n):
    print(f"\n--- Pessoa {i+1} ---")
    nome = input("Nome: ")
    idade = int(input("Idade: "))
    sexo = input("Sexo (M/F): ").strip().upper()

    soma_idade += idade

    if sexo == 'F' and idade < 20:
        mulheres_menos_20 += 1

media_idade = soma_idade / n if n > 0 else 0

print(f"\nA média de idade do grupo é: {media_idade:.2f} anos.")
print(f"Mulheres com menos de 20 anos: {mulheres_menos_20}")
```

### Questão 7

```python
lista_receita = ['Farinha de trigo', 'Ovos', 'Leite', 'Açúcar', 'Fermento', 'Manteiga']

pessoa_1 = {'Ovos', 'Leite'}
pessoa_2 = {'Farinha de trigo'}

conjunto_receita = set(lista_receita)

tem_em_casa = pessoa_1 | pessoa_2

falta_comprar = conjunto_receita - tem_em_casa

print(f"Ingredientes que faltam comprar: {falta_comprar}")
```

### Questão 8

```python
estoque = []

for i in range(3):
    print(f"\n--- Produto {i+1} ---")
    nome = input("Nome do produto: ")
    preco = float(input("Preço: R$ "))
    quantidade = int(input("Quantidade em estoque: "))

    produto = {'nome': nome, 'preco': preco, 'quantidade': quantidade}
    estoque.append(produto)

print("\n--- Relatório de Estoque ---")
for prod in estoque:
    valor_total = prod['preco'] * prod['quantidade']
    print(f"Produto: {prod['nome']} | Valor Total em Estoque: R$ {valor_total:.2f}")
```
